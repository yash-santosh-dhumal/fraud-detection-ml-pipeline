```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
```

```python
#read the dataset and display the first 5 rows
df = pd.read_csv("../data/creditcard.csv")
df.head()
```

```python
# shape of the dataset
df.shape
```

```python
#getting column names, data types and non-null counts and information about the dataset
df.info()
```

```python
#checking for missing values in the dataset
df.isnull().sum()
```

```python
#getting statistical summary of the dataset(count, mean, std, min, 25%, 50%, 75% and max)
df.describe()
```

```python
#analyzing fraudulent and non-fraudulent transactions in the dataset where 0 is non-fraudulent and 1 is fraudulent
df['Class'].value_counts()
```

```python
#visualixing the distrubution of fraud vs normal transactions in the dataset
sns.countplot(x='Class', data=df)
plt.title('Fraudulent vs Normal Transactions')
plt.show()
```

```python
#Compare fraud vs normal transcations based on the amount of the transaction
plt.figure(figsize=(8, 5))
sns.histplot(df['Amount'], bins=50)
plt.title('Distribution of Transaction Amounts')
plt.show()
```

```python
#creating a new feature 'Hour' by converting the 'Time' feature from seconds to hours
df['Hour'] = df['Time'] / 3600
```

```python
plt.figure(figsize=(10,5))
sns.histplot(data=df, x='Hour', hue='Class', bins=50)
plt.title('Distribution of Transactions by Hour')
plt.show()
```

```python
#correlation heatmap to visualize the correlation between the features in the dataset

plt.figure(figsize=(14,10))
sns.heatmap(df.corr(), cmap='coolwarm')
plt.title('Correlation Heatmap matrix ')
plt.show()
```

```python
#sorting the correlation values of the features with respect to the target variable 'Class' in descending order
corr = df.corr()

corr['Class'].sort_values(ascending=False)
```

2. Fraud transactions represent only 0.17% of the dataset,
   indicating a severe class imbalance.

3. No missing values were found in the dataset.

4. Transaction amounts show a highly skewed distribution.

5. Features V14, V17, and V12 show relatively strong correlation

6. Fraud transactions appear irregular across time, suggesting
